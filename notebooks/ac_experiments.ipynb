## The ten A2C experiments

model, history, config and plot under models/ac/exp<nn>/.

```python
!"{sys.executable}" training/actor_critic_training.py --exp 1
```

Experiment 2: higher learning rate.

```python
!"{sys.executable}" training/actor_critic_training.py --exp 2 --lr 3e-3
```

Experiment 3: lower learning rate.

```python
!"{sys.executable}" training/actor_critic_training.py --exp 3 --lr 2e-4
```

Experiment 4: longer horizon.

```python
!"{sys.executable}" training/actor_critic_training.py --exp 4 --gamma 0.995
```

Experiment 5: shorter horizon.

```python
!"{sys.executable}" training/actor_critic_training.py --exp 5 --gamma 0.95
```

Experiment 6: more exploration.

```python
!"{sys.executable}" training/actor_critic_training.py --exp 6 --entropy 0.05
```

Experiment 7: no entropy bonus.

```python
!"{sys.executable}" training/actor_critic_training.py --exp 7 --entropy 0.0
```

Experiment 8: bigger rollout.

```python
!"{sys.executable}" training/actor_critic_training.py --exp 8 --frames-per-batch 1000
```

Experiment 9: shorter GAE lambda.

```python
!"{sys.executable}" training/actor_critic_training.py --exp 9 --lmbda 0.9
```

Experiment 10: bigger network.
