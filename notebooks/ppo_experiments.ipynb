## The ten PPO experiments
Each run changes one setting from the baseline: learning rate, the clip range, the number of
update epochs per batch, the entropy bonus, gamma for the horizon, and the network size. Every
run saves its model, history, config and plot under models/ppo/exp<nn>/.

Experiment 1: baseline.

```python
!"{sys.executable}" training/ppo_training.py --exp 1
```

Experiment 2: higher learning rate.

```python
!"{sys.executable}" training/ppo_training.py --exp 2 --lr 1e-3
```

Experiment 3: lower learning rate.

```python
!"{sys.executable}" training/ppo_training.py --exp 3 --lr 1e-4
```

Experiment 4: tighter clip.

```python
!"{sys.executable}" training/ppo_training.py --exp 4 --clip 0.1
```

Experiment 5: looser clip.

```python
!"{sys.executable}" training/ppo_training.py --exp 5 --clip 0.3
```

Experiment 6: more update epochs.

```python
!"{sys.executable}" training/ppo_training.py --exp 6 --epochs 8
```

Experiment 7: fewer update epochs.

```python
!"{sys.executable}" training/ppo_training.py --exp 7 --epochs 2
```

Experiment 8: more exploration.

```python
!"{sys.executable}" training/ppo_training.py --exp 8 --entropy 0.05
```

Experiment 9: longer horizon.

```python
!"{sys.executable}" training/ppo_training.py --exp 9 --gamma 0.995
```

Experiment 10: bigger network.

```python
!"{sys.executable}" training/ppo_training.py --exp 10 --hidden 256
```
