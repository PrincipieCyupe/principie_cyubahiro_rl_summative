## The ten DQN experiments
Each run changes one setting from the baseline: exploration through epsilon decay, replay
through the buffer size, the target network through the sync interval, and learning rate,
gamma and batch size. Every run saves its model, history, config and plot under
models/dqn/exp<nn>/.

Experiment 1: baseline.

```python
!"{sys.executable}" training/dqn_training.py --exp 1
```

Experiment 2: higher learning rate.

```python
!"{sys.executable}" training/dqn_training.py --exp 2 --lr 5e-3
```

Experiment 3: lower learning rate.

```python
!"{sys.executable}" training/dqn_training.py --exp 3 --lr 2.5e-4
```

Experiment 4: longer horizon.

```python
!"{sys.executable}" training/dqn_training.py --exp 4 --gamma 0.995
```

Experiment 5: shorter horizon.

```python
!"{sys.executable}" training/dqn_training.py --exp 5  --gamma 0.95
```

Experiment 6: bigger batch.

```python
!"{sys.executable}" training/dqn_training.py --exp 6 --batch-size 128 --gamma 0.995
```

Experiment 7: smaller replay buffer.

```python
!"{sys.executable}" training/dqn_training.py --exp 7 --buffer-size 10000 --gamma 0.995
```

Experiment 8: faster target sync.

```python
!"{sys.executable}" training/dqn_training.py --exp 8 --target-update 100 --gamma 0.995
```

Experiment 9: slower target sync.

```python
!"{sys.executable}" training/dqn_training.py --exp 9 --target-update 2000 --gamma 0.995
```

Experiment 10: slower epsilon decay, more exploration.

```python
!"{sys.executable}" training/dqn_training.py --exp 10 --eps-decay 40000 --gamma 0.995
```
