## The ten REINFORCE experiments
Each run changes one setting from the baseline: learning rate, gamma for the horizon, the
rollout size (steps collected before each update), the entropy bonus that keeps the policy
exploring, the network size, and lambda, where 1.0 is pure Monte Carlo and a bit lower adds
some bootstrap. Every run saves its model, history, config and plot under
models/reinforce/exp<nn>/.

Experiment 1: baseline.

```python
!"{sys.executable}" training/reinforce_training.py --exp 1 
# using the default values set in reinforce_training.py
```

Experiment 2: higher learning rate.

```python
!"{sys.executable}" training/reinforce_training.py --exp 2 --lr 5e-3
```

Experiment 3: lower learning rate.

```python
!"{sys.executable}" training/reinforce_training.py --exp 3 --lr 2.5e-4
```

Experiment 4: longer horizon.

```python
!"{sys.executable}" training/reinforce_training.py --exp 4 --gamma 0.995
```

Experiment 5: shorter horizon.

```python
!"{sys.executable}" training/reinforce_training.py --exp 5 --gamma 0.95
```

Experiment 6: bigger rollout and increased gamma

```python
!"{sys.executable}" training/reinforce_training.py --exp 6 --frames-per-batch 2000 --gamma 0.997
```

Experiment 7: more exploration.

```python
!"{sys.executable}" training/reinforce_training.py --exp 7 --entropy 0.05
```

Experiment 8: no entropy bonus.

```python
!"{sys.executable}" training/reinforce_training.py --exp 8 --entropy 0.0
```

Experiment 9: bigger network.

```python
!"{sys.executable}" training/reinforce_training.py --exp 9 --hidden 256
```

Experiment 10: some bootstrap.

```python
!"{sys.executable}" training/reinforce_training.py --exp 10 --lmbda 0.95
```
